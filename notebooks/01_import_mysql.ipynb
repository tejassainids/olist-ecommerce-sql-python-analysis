```python
import pandas as pd
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine

load_dotenv()

user = os.getenv("MYSQL_USER")
password = os.getenv("MYSQL_PASSWORD")
host = os.getenv("MYSQL_HOST")
database = os.getenv("MYSQL_DATABASE")

engine = create_engine(
    f"mysql+mysqlconnector://{user}:{password}@{host}/{database}"
)
```

```python
connection = engine.connect()
print("Connected Successfully!")
connection.close()
```

```python
DATA_PATH = "/Users/tejasmac/Data Analytics Portfolio/EDA_Projects/olist-ecommerce-sql-python-analysis/data"
```

```python
customers = pd.read_csv(
    f"{DATA_PATH}/olist_customers_dataset.csv"
)

customers.head()
```

```python
#customers
customers.to_sql(
    "customers",
    con=engine,
    if_exists="append",
    index=False
)

print("Customers imported successfully!")
```

```python
def import_csv(file_name, table_name):
    #Read a CSV file and append its contents to a MySQL table.
    df = pd.read_csv(f"{DATA_PATH}/{file_name}")

    df.to_sql(
        table_name,
        con=engine,
        if_exists="append",
        chunksize=5000,
        index=False
    )

    print(f"✅ {table_name} imported successfully! ({len(df)} rows)")
```

```python
#products
products = pd.read_csv(f"{DATA_PATH}/olist_products_dataset.csv")

products.rename(columns={
    "product_name_lenght": "product_name_length",
    "product_description_lenght": "product_description_length",
    "product_weight_g": "product_weight",
    "product_length_cm": "product_length",
    "product_height_cm": "product_height",
    "product_width_cm": "product_width"
}, inplace=True)

products.to_sql(
    "products",
    con=engine,
    if_exists="append",
    index=False
)

print("Products imported successfully!")
```
